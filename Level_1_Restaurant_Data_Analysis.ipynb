# Level 1 - Task 3: Price Range Analysis

## Objective

Analyze the distribution of restaurants across different price ranges and identify the most common price range.

```python
price_counts = df['Price range'].value_counts().sort_index()

price_counts
```

```python
price_percentage = (price_counts / price_counts.sum()) * 100

price_percentage.round(2)
```

```python
plt.figure(figsize=(8,5))

price_counts.plot(kind='bar')

plt.title("Distribution of Restaurants by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=0)

plt.show()
```

## Observation

- Price Range 1 has the highest number of restaurants (4438), accounting for 46.51% of the dataset.
- Price Range 2 is the second most common with 3113 restaurants (32.62%).
- Price Range 3 has 1405 restaurants (14.72%).
- Price Range 4 has the lowest number of restaurants (586), representing only 6.14% of the dataset.
- Most restaurants belong to the lower price ranges (1 and 2).

## Conclusion

The analysis shows that most restaurants are affordable, with Price Range 1 being the most common. As the price range increases, the number of restaurants decreases. This indicates that lower-priced restaurants are more common in the dataset.

## Objective

Analyze online delivery services by calculating the percentage of restaurants offering online delivery and compare the average ratings of restaurants with and without online delivery.

```python
import matplotlib.pyplot as plt

# Count of online delivery availability
online_delivery = df['Has Online delivery'].value_counts()

# Bar chart
plt.figure(figsize=(6,4))
online_delivery.plot(kind='bar')

plt.xlabel("Online Delivery Availability")
plt.ylabel("Number of Restaurants")
plt.title("Restaurants with Online Delivery Facility")

plt.xticks(rotation=0)
plt.show()
```

Observation:

Out of these, 7091 restaurants do not provide online delivery services.
Only 2451 restaurants provide online delivery facilities.
The majority of restaurants are not offering online delivery services.

Conclusion:

The analysis shows that online delivery adoption is low among restaurants. Around 74% of restaurants do not provide online delivery, while only 26% offer this facility. Restaurants can improve customer reach and sales by adopting online delivery services.
